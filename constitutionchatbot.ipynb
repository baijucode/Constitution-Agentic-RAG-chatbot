```python
import requests
import json

def search_tavily_for_constitution(legal_query: str) -> str:
    """
    Searches the web via Tavily API for specific constitutional queries 
    and returns a combined string of the most relevant text snippets.
    """
    TAVILY_API_KEY = "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"
    url = "https://api.tavily.com/search"
    
    headers = {
        "Authorization": f"Bearer {TAVILY_API_KEY}",
        "Content-Type": "application/json"
    }
    
    # We change 'query' to be dynamic, and increase max_results slightly for better legal context
    payload = {
        "query": legal_query,
        "topic": "general",
        "search_depth": "advanced", # "advanced" works better for complex legal/constitutional topics
        "max_results": 3,
        "include_answer": False,
        "include_raw_content": False
    }
    
    try:
        response = requests.post(url, headers=headers, json=payload)
        if response.status_code == 200:
            results = response.json().get("results", [])
            
            # Combine the titles and snippets from the web results into one clean text string
            web_context = ""
            for i, res in enumerate(results, 1):
                web_context += f"[Web Source {i}]: {res.get('title')}\n{res.get('content')}\n\n"
            
            return web_context if web_context else "No matching web documents found."
        else:
            return f"Error fetching web context (Status {response.status_code})"
            
    except Exception as e:
        return f"Tavily search connection failed: {e}"
```

```python
def chatbot(message, history):
    try:
        # 1. Fetch Local Context from PDF Index
        local_docs = vector_store.similarity_search(message, k=3)
        local_context = "\n\n".join(doc.page_content for doc in local_docs)

        # 2. Fetch Live Context from the Web using new Tavily function
        web_context = search_tavily_for_constitution(message)

        # 3. Combine both sources together into one massive context variable
        combined_context = f"--- LOCAL PDF EXTRACTS ---\n{local_context}\n\n--- LIVE WEB ANALYSIS ---\n{web_context}"

        # 4. Formatting existing prompt template
        messages = prompt.format_prompt(
            chat_history=history,
            context=combined_context,
            question=message
        ).to_messages()

        # 5. Invoking local LM Studio model
        response = chat_model.invoke(messages)
        return response.content

    except Exception as e:
        return f"⚠️ Something went wrong while generating a response: {e}"
```

```python
def generate_search_query(user_message, history) -> str:
    """Uses the chat model to turn conversational follow-ups into sharp search queries."""
    if not history:
        return user_message  # If it's the first message, the raw message is fine.
        
    query_prompt = f"""
    Given the following conversation history and a follow-up question, rephrase the follow-up question to be a standalone search query for a legal search engine.
    
    History: {history[-2:]}
    Follow-up: {user_message}
    Standalone Query:"""
    
    try:
        response = chat_model.invoke(query_prompt)
        return response.content.strip().strip('"')
    except Exception:
        return user_message  # Fallback to original message if LLM fails
```

```python
from strands import Agent
from strands.models.openai import OpenAIModel

BASE_URL = 'http://localhost:1234/v1' # Keep this localhost to avoid connection errors
API_KEY = 'lm-studio'
CHAT_MODEL = 'qwen2.5-vl-3b-instruct'

local_model = OpenAIModel(
    model_id=CHAT_MODEL,
    client_args={
        "base_url": BASE_URL,
        "api_key": API_KEY
    }
)
```

```python
from strands import Agent

# 1. Turn  FAISS database into a function (Tool) the agent can use
def query_nepal_constitution(legal_question: str) -> str:
    """Use this tool to search the local FAISS database for official clauses, 
    articles, and raw text inside the Constitution of Nepal. Always use this first."""
    docs = vector_store.similarity_search(legal_question, k=4)
    return "\n\n".join(doc.page_content for doc in docs)

# 2. Turn  Tavily search into a function (Tool) the agent can use
def live_web_search(query: str) -> str:
    """Use this tool to search the internet for live legal commentary, supreme court 
    interpretations, explanations, or recent legal news regarding Nepal."""
    return search_tavily_for_constitution(query)

# 3. Create the Agent with a clean persona prompt and hand it the tools
agent = Agent(
    model=local_model, 
    tools=[query_nepal_constitution, live_web_search],
    system_prompt=(
        "You are an expert AI Legal Assistant specializing in the Constitution of Nepal. "
        "Use your local database tool for the raw constitutional text, and the web search tool "
        "for explanations or commentary. The local database text is the ultimate ground truth. "
        "Always cite the Article numbers or sources clearly in your final answer."
    )
)
```
